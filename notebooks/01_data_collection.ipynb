```python
def get_ons_monthly_series(series_id, dataset="mgdp"):
    url = f"https://www.ons.gov.uk/economy/grossdomesticproductgdp/timeseries/{series_id}/{dataset}/data"
    response = requests.get(url)
    response.raise_for_status()

    data = response.json()
    df = pd.DataFrame(data["months"])

    df["value"] = pd.to_numeric(df["value"], errors="coerce")

    return df
```

```python
services_test = get_ons_monthly_series("eczg")

services_test.head()
```

```python
construction = get_ons_monthly_series("ecze")

construction.head()
```

```python
construction.tail()
```

```python
construction.to_csv("../data/raw/construction_output.csv",index=False)
```
