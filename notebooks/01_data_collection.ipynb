```python
def get_ons_monthly_series(series_id, dataset="mgdp"):
    url = f"https://www.ons.gov.uk/economy/grossdomesticproductgdp/timeseries/{series_id}/{dataset}/data"
    response = requests.get(url)
    response.raise_for_status()

    data = response.json()
    df = pd.DataFrame(data["months"])

    df["value"] = pd.to_numeric(df["value"], errors="coerce")

    return df
```

```python
services_test = get_ons_monthly_series("eczg")

services_test.head()
```

```python
construction = get_ons_monthly_series("ecze")

construction.head()
```

```python
construction.tail()
```

```python
construction.to_csv("../data/raw/construction_output.csv",index=False)
```

```python
oecd_url = (
    "https://sdmx.oecd.org/public/rest/data/"
    "OECD.SDD.STES,DSD_STES@DF_CLI/"
    "GBR.M.LI...AA...H"
    "?startPeriod=2006-01"
    "&dimensionAtObservation=AllDimensions"
    "&format=csvfilewithlabels"
)

oecd_response = requests.get(oecd_url)

print(oecd_response.status_code)
print(oecd_response.url)
print(oecd_response.text[:300])
```

```python
from io import StringIO

oecd_cli = pd.read_csv(StringIO(oecd_response.text))

oecd_cli.head()
```

```python
print(oecd_cli.columns.tolist())
print(oecd_cli.shape)
```

```python
cli = oecd_cli[["TIME_PERIOD", "OBS_VALUE"]].copy()

cli.columns = ["date", "value"]

cli.head()
```
